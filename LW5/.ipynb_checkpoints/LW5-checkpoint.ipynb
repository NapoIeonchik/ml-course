```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, validation_curve
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.ensemble import GradientBoostingClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
```

```python
df = pd.read_csv("adult.csv", delimiter=',')
```

```python
data_cleaned = df.replace('?', np.nan)
data_cleaned['income'] = data_cleaned['income'].apply(lambda x: 1 if x == '>50K' else 0)
data_dropped = data_cleaned.dropna()
```

```python
data_cleaned = data_dropped.drop(columns=['educational-num'])
```

```python
for col in data_cleaned.select_dtypes(include=np.number).columns:
    df_encoded = pd.get_dummies(data_cleaned, drop_first=True)
```

```python
correlation_matrix = df_encoded.corr()
correlation_with_target = correlation_matrix['income'].sort_values(ascending=False)
print(correlation_with_target)

selected_features = correlation_with_target[correlation_with_target.abs() > 0.2].index.tolist()
selected_features.remove('income')  
print("Вибрані ознаки для моделі:", selected_features)
```

```python
X_1 = df_encoded[['marital-status_Married-civ-spouse', 'age', 'hours-per-week', 'capital-gain', 'gender_Male', 'occupation_Exec-managerial', 'relationship_Own-child', 'marital-status_Never-married']]
y_1 = df_encoded['income']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_1, y_1, test_size=0.3, random_state=42, stratify=y_1)
```

```python
# AdaBoost
base_model = DecisionTreeClassifier(max_depth=1, random_state=42)
ada_model = AdaBoostClassifier(estimator=base_model, random_state=42)

param_range_estimators = np.arange(10, 210, 20)
param_range_lr = np.logspace(-3, 0, 4)

train_scores_estimators, test_scores_estimators = validation_curve(
    ada_model, X_train, y_train,
    param_name="n_estimators", param_range=param_range_estimators,
    scoring="accuracy", cv=5
)
train_scores_lr, test_scores_lr = validation_curve(
    ada_model, X_train, y_train,
    param_name="learning_rate", param_range=param_range_lr,
    scoring="accuracy", cv=5
)

train_mean_estimators = np.mean(train_scores_estimators, axis=1)
train_std_estimators = np.std(train_scores_estimators, axis=1)
test_mean_estimators = np.mean(test_scores_estimators, axis=1)
test_std_estimators = np.std(test_scores_estimators, axis=1)

train_mean_lr = np.mean(train_scores_lr, axis=1)
train_std_lr = np.std(train_scores_lr, axis=1)
test_mean_lr = np.mean(test_scores_lr, axis=1)
test_std_lr = np.std(test_scores_lr, axis=1)

fig, axes = plt.subplots(1, 2, figsize=(15, 6))
axes[0].plot(param_range_estimators, train_mean_estimators, label="Training score", color="blue")
axes[0].plot(param_range_estimators, test_mean_estimators, label="Cross-validation score", color="green")
axes[0].fill_between(param_range_estimators, train_mean_estimators - train_std_estimators, train_mean_estimators + train_std_estimators, alpha=0.2, color="blue")
axes[0].fill_between(param_range_estimators, test_mean_estimators - test_std_estimators, test_mean_estimators + test_std_estimators, alpha=0.2, color="green")
axes[0].set_title("Validation Curve with AdaBoost (n_estimators)")
axes[0].set_xlabel("Number of Estimators")
axes[0].set_ylabel("Accuracy")
axes[0].legend(loc="best")
axes[0].grid()

axes[1].plot(param_range_lr, train_mean_lr, label="Training score", color="blue")
axes[1].plot(param_range_lr, test_mean_lr, label="Cross-validation score", color="green")
axes[1].fill_between(param_range_lr, train_mean_lr - train_std_lr, train_mean_lr + train_std_lr, alpha=0.2, color="blue")
axes[1].fill_between(param_range_lr, test_mean_lr - test_std_lr, test_mean_lr + test_std_lr, alpha=0.2, color="green")
axes[1].set_title("Validation Curve with AdaBoost (learning_rate)")
axes[1].set_xlabel("Learning Rate")
axes[1].set_ylabel("Accuracy")
axes[1].legend(loc="best")
axes[1].grid()
plt.tight_layout()
plt.show()
```

```python
gb_model = GradientBoostingClassifier(random_state=42, max_depth=3)

param_range_estimators = np.arange(50, 310, 50)
param_range_lr = np.logspace(-3, 0, 4)

train_scores_estimators, test_scores_estimators = validation_curve(
    gb_model, X_train, y_train,
    param_name="n_estimators", param_range=param_range_estimators,
    scoring="accuracy", cv=5
)
train_scores_lr, test_scores_lr = validation_curve(
    gb_model, X_train, y_train,
    param_name="learning_rate", param_range=param_range_lr,
    scoring="accuracy", cv=5
)

train_mean_estimators = np.mean(train_scores_estimators, axis=1)
train_std_estimators = np.std(train_scores_estimators, axis=1)
test_mean_estimators = np.mean(test_scores_estimators, axis=1)
test_std_estimators = np.std(test_scores_estimators, axis=1)

train_mean_lr = np.mean(train_scores_lr, axis=1)
train_std_lr = np.std(train_scores_lr, axis=1)
test_mean_lr = np.mean(test_scores_lr, axis=1)
test_std_lr = np.std(test_scores_lr, axis=1)

fig, axes = plt.subplots(1, 2, figsize=(15, 6))
axes[0].plot(param_range_estimators, train_mean_estimators, label="Training score", color="blue")
axes[0].plot(param_range_estimators, test_mean_estimators, label="Cross-validation score", color="green")
axes[0].fill_between(param_range_estimators, train_mean_estimators - train_std_estimators, train_mean_estimators + train_std_estimators, alpha=0.2, color="blue")
axes[0].fill_between(param_range_estimators, test_mean_estimators - test_std_estimators, test_mean_estimators + test_std_estimators, alpha=0.2, color="green")
axes[0].set_title("Validation Curve with GradientBoosting (n_estimators)")
axes[0].set_xlabel("Number of Estimators")
axes[0].set_ylabel("Accuracy")
axes[0].legend(loc="best")
axes[0].grid()

axes[1].plot(param_range_lr, train_mean_lr, label="Training score", color="blue")
axes[1].plot(param_range_lr, test_mean_lr, label="Cross-validation score", color="green")
axes[1].fill_between(param_range_lr, train_mean_lr - train_std_lr, train_mean_lr + train_std_lr, alpha=0.2, color="blue")
axes[1].fill_between(param_range_lr, test_mean_lr - test_std_lr, test_mean_lr + test_std_lr, alpha=0.2, color="green")
axes[1].set_title("Validation Curve with GradientBoosting (learning_rate)")
axes[1].set_xlabel("Learning Rate")
axes[1].set_ylabel("Accuracy")
axes[1].legend(loc="best")
axes[1].grid()
plt.tight_layout()
plt.show()
```

```python
xgb_model = XGBClassifier(random_state=42, eval_metric="logloss")
lgbm_model = LGBMClassifier(random_state=42)

param_range = np.arange(50, 310, 50)

xgb_train_scores, xgb_test_scores = validation_curve(
    xgb_model, X_train, y_train,
    param_name="n_estimators", param_range=param_range,
    scoring="accuracy", cv=5
)
lgbm_train_scores, lgbm_test_scores = validation_curve(
    lgbm_model, X_train, y_train,
    param_name="n_estimators", param_range=param_range,
    scoring="accuracy", cv=5
)

xgb_train_mean = np.mean(xgb_train_scores, axis=1)
xgb_train_std = np.std(xgb_train_scores, axis=1)
xgb_test_mean = np.mean(xgb_test_scores, axis=1)
xgb_test_std = np.std(xgb_test_scores, axis=1)

lgbm_train_mean = np.mean(lgbm_train_scores, axis=1)
lgbm_train_std = np.std(lgbm_train_scores, axis=1)
lgbm_test_mean = np.mean(lgbm_test_scores, axis=1)
lgbm_test_std = np.std(lgbm_test_scores, axis=1)

plt.figure(figsize=(14, 6))
# XGBoost
plt.subplot(1, 2, 1)
plt.plot(param_range, xgb_train_mean, label="Training score", color="blue")
plt.plot(param_range, xgb_test_mean, label="Cross-validation score", color="green")
plt.fill_between(param_range, xgb_train_mean - xgb_train_std, xgb_train_mean + xgb_train_std, alpha=0.2, color="blue")
plt.fill_between(param_range, xgb_test_mean - xgb_test_std, xgb_test_mean + xgb_test_std, alpha=0.2, color="green")
plt.title("validation curve with XGBoost")
plt.xlabel("nums of estimators")
plt.ylabel("accuracy")
plt.legend(loc="best")
plt.grid()

# LightGBM
plt.subplot(1, 2, 2)
plt.plot(param_range, lgbm_train_mean, label="Training score", color="blue")
plt.plot(param_range, lgbm_test_mean, label="Cross-validation score", color="green")
plt.fill_between(param_range, lgbm_train_mean - lgbm_train_std, lgbm_train_mean + lgbm_train_std, alpha=0.2, color="blue")
plt.fill_between(param_range, lgbm_test_mean - lgbm_test_std, lgbm_test_mean + lgbm_test_std, alpha=0.2, color="green")
plt.title("validation curve with LightGBM")
plt.xlabel("nums of estimators")
plt.ylabel("accuracy")
plt.legend(loc="best")
plt.grid()

plt.tight_layout()
plt.show()
```

```python
from xgboost import plot_importance

xg_model = XGBClassifier(random_state=42)
xg_model.fit(X_train, y_train)

plt.figure(figsize=(10, 6))
plot_importance(xg_model, importance_type='weight', max_num_features=10, title='Feature Importance (XGBoost) by Weight')
plt.show()

plt.figure(figsize=(10, 6))
plot_importance(xg_model, importance_type='gain', max_num_features=10, title='Feature Importance (XGBoost) by Gain')
plt.show()
```

```python
import lightgbm as lgb

lgb_model = LGBMClassifier(random_state=42)
lgb_model.fit(X_train, y_train)

plt.figure(figsize=(10, 6))
lgb.plot_importance(lgb_model, importance_type='split', max_num_features=10, title='Feature Importance (LightGBM) by Split')
plt.show()

plt.figure(figsize=(10, 6))
lgb.plot_importance(lgb_model, importance_type='gain', max_num_features=10, title='Feature Importance (LightGBM) by Gain')
plt.show()
```
